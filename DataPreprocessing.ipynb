```python
import pandas as pd
import numpy as np

df = pd.read_csv("collegePlace_25.csv")

print(df.head())

print(df.info())

print(df.isnull().sum())

print(df.describe())

# Handle Missing Values

# Numerical Columns
num_cols = ['Age', 'Internships', 'CGPA',
            'Hostel', 'HistoryOfBacklogs']

for col in num_cols:
    df[col].fillna(df[col].mean(), inplace=True)

# Categorical Columns
cat_cols = ['Gender', 'Branch']

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("\nAfter Handling Missing Values:")
print(df.head())

# Label Encoding (Gender)

gender_mapping = {
    'Male': 0,
    'Female': 1
}

df['Gender'] = df['Gender'].map(gender_mapping)

# One-Hot Encoding (Branch)

df = pd.get_dummies(df, columns=['Branch'])

print("\nAfter Encoding:")
print(df.head())

# Standardization (Z-score)

feature_cols = df.drop('PlacedOrNot', axis=1).columns

for col in feature_cols:
    mean = df[col].mean()
    std = df[col].std()

    df[col] = (df[col] - mean) / std

print("\nAfter Standardization:")
print(df.head())

# Train-Test Split (80:20)

df = df.sample(frac=1, random_state=42)

train_size = int(0.8 * len(df))

train_data = df[:train_size]
test_data = df[train_size:]

print("\nTraining Data Shape:")
print(train_data.shape)

print("\nTesting Data Shape:")
print(test_data.shape)

print("\nTraining Data:")
print(train_data.head())

print("\nTesting Data:")
print(test_data.head())
```
