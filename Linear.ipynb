```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("mobile_screen_time_battery_usage.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
x = df['Screen_Time_Hours'].values
y = df['Battery_Used_Percent'].values

w = 0
b = 0
lr = 0.01
epochs = 5000

for i in range(epochs):
    y_pred = w * x + b

    error = y - y_pred

    dw = (-2 * x * error).mean()
    db = (-2 * error).mean()

    w = w - lr * dw
    b = b - lr * db

print("Weight (w):", w)
print("Bias (b):", b)

y_pred = w * x + b

plt.scatter(x, y)
plt.plot(x, y_pred, color='red')

plt.xlabel("Screen_Time_Hours")
plt.ylabel("Battery_Used_Percent")
plt.title("Linear Regression using Gradient Descent")

plt.show()
```
